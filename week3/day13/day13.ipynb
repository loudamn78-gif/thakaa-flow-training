### 1. لماذا نرسم؟
في رباعية أنسكومب (Anscombe's quartet) فيه 4 مجموعات بيانات لها نفس المتوسط ونفس الانحراف المعياري ونفس الارتباط تقريبًا، لكن لما رسمتها طلعت أشكالها مختلفة تمامًا. هذا يعني لمحلل البيانات إنه ما يكفي يعتمد على الأرقام بس، لازم يرسم البيانات عشان يشوف شكلها الحقيقي (قيم شاذة، تجمعات، علاقة منحنية) وما ينخدع بالملخصات.

مثال من حياتي: معدلي التراكمي رقم واحد، لكن ممكن يخفي إن درجاتي في مادة كانت ضعيفة جدًا وفي مادة ثانية ممتازة، أو إن مستواي كان يتحسن كل فصل. الرقم الواحد يخفي القصة، والرسمة توضحها.

### 2. اختر الرسمة

| السؤال | الرسمة؟ | لماذا؟ |
|---|---|---|
| كيف تغيرت مبيعات المقهى خلال 12 شهرًا؟ | Line (`plt.plot`) | لأن البيانات مرتبطة بالزمن، والخط يوضح الاتجاه صعودًا ونزولًا عبر الأشهر. |
| أي فرع باع أكثر هذا الشهر من 5 فروع؟ | Bar (`plt.bar`) | لأني أقارن قيمة واحدة لكل فئة (الفرع)، والأعمدة تسهّل معرفة الأعلى. |
| كيف تتوزع أعمار 500 عميل؟ | Histogram (`plt.hist`) | لأن الأعمار قيم رقمية كثيرة وأبي أعرف توزيعها وأين تتركز. |
| هل يزيد الإيجار كلما زادت مساحة الشقة؟ | Scatter (`plt.scatter`) | لأني أبحث عن علاقة بين متغيرين رقميين (المساحة والإيجار). |
| هل في رواتب الموظفين قيم شاذة؟ | Histogram (`plt.hist`) | لأنه يبيّن توزيع الرواتب، وأي قيمة بعيدة عن الباقي تظهر منفصلة في الطرف. |

### 3. أجزاء الرسمة
- **Figure** هي الصفحة أو اللوحة الكاملة اللي فيها كل شيء، و**Axes** هي منطقة الرسم نفسها اللي فيها المحورين والبيانات. ممكن يكون في الـ Figure الواحد أكثر من Axes (مثل `subplots`).
- **العنوان (title):** يقول عن ماذا الرسمة. **عنوانا المحورين (xlabel / ylabel):** يوضحان ماذا يمثل كل محور ووحدته. **الـ Legend:** يوضح أي لون أو خط يمثل أي مجموعة بيانات.
- لو أرسلت رسمة بلا أي منها لمديري، ما راح يعرف ماذا تمثل الأرقام ولا وحدتها ولا أي خط هو أي شيء، فيخمّن أو يفسّرها غلط وتصير الرسمة بلا فائدة.

### 4. رسمة مضلّلة
الخدعة إن محور Y ما يبدأ من الصفر، يبدأ من رقم قريب من 94% مثلًا، فالفرق الصغير (96% مقابل 94%) يبان كأن العمود الجديد أكبر بكثير من القديم، والقديم يبدو ضعيف جدًا.

## Task 2: رسومات المقهى


```python
import matplotlib.pyplot as plt

days = ["Sat", "Sun", "Mon", "Tue", "Wed", "Thu", "Fri"]
latte = [42, 38, 35, 40, 44, 58, 61]
espresso = [30, 28, 33, 31, 29, 35, 22]
drinks = ["Latte", "Espresso", "Mocha", "Tea"]
ratings = [5, 4, 4, 5, 3, 5, 4, 2, 5, 4, 4, 5, 3, 4, 5, 5, 1, 4, 4, 5,
           3, 4, 5, 4, 2, 5, 4, 4, 3, 5, 4, 5, 5, 4, 3, 4, 5, 4, 4, 5]
temp = [18, 22, 25, 27, 30, 33, 35, 38, 40, 41]
iced = [5, 9, 12, 15, 20, 26, 30, 35, 39, 38]
```

### 1. Line: مبيعات Latte و Espresso

```python
plt.plot(days, latte, marker="o", label="Latte")
plt.plot(days, espresso, marker="s", label="Espresso")
plt.title("Daily Sales: Latte vs Espresso")
plt.xlabel("Day of the week")
plt.ylabel("Cups sold")
plt.legend()
plt.show()
```

*** مبيعات الـ Latte ترتفع بوضوح نهاية الأسبوع، وأعلى يوم لها **الجمعة (Fri)** بـ 61 كوب، بينما الـ Espresso تبقى شبه ثابتة بين 28 و35 تقريبًا.

**ماذا حدث لـ Espresso يوم الجمعة؟** نزلت لأقل قيمة في الأسبوع (22 كوب) بنفس اليوم اللي وصل فيه الـ Latte لأعلى مبيعاته.

### 2. Bar: إجمالي الأسبوع لكل مشروب

```python
totals = [sum(latte), sum(espresso), 145, 96]   # Mocha = 145, Tea = 96
print(totals)

plt.bar(drinks, totals, color="saddlebrown")
for i, v in enumerate(totals):
    plt.text(i, v + 3, str(v), ha="center")
plt.title("Total Weekly Sales per Drink")
plt.xlabel("Drink")
plt.ylabel("Cups sold in the week")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("insurance.csv")
df.head()
```

### 1. Histogram للتكاليف مع خطي المتوسط والوسيط

```python
mean_c = df["charges"].mean()
median_c = df["charges"].median()

plt.hist(df["charges"], bins=30, edgecolor="black")
plt.axvline(mean_c, color="red", linestyle="--", label=f"Mean = {mean_c:,.0f}")
plt.axvline(median_c, color="green", linestyle="-", label=f"Median = {median_c:,.0f}")
plt.title("Distribution of Medical Charges")
plt.xlabel("Charges")
plt.ylabel("Number of people")
plt.legend()
plt.show()
```

** التوزيع ملتوٍ لليمين (right-skewed): أغلب الناس تكاليفهم منخفضة، وفي ذيل طويل لتكاليف عالية. المتوسط أكبر من الوسيط لأن القيم العالية تسحبه لفوق، وهذا يعني إن الوسيط أصدق تمثيل للشخص "العادي".

### 2. Scatter: العمر مقابل التكاليف، بلونين (مدخن / غير مدخن)

```python
colors = df["smoker"].map({"yes": "red", "no": "blue"})
plt.scatter(df["age"], df["charges"], c=colors, alpha=0.6)
plt.scatter([], [], c="red", label="Smoker")
plt.scatter([], [], c="blue", label="Non-smoker")
plt.title("Age vs Charges by Smoking Status")
plt.xlabel("Age")
plt.ylabel("Charges")
plt.legend()
plt.show()
```

**كم «شريطًا» تراه؟** أشوف **3 أشرطة** تقريبًا، كلها تصعد مع العمر.

**من في الشريط الأعلى؟** المدخنون (النقاط الحمراء) هم الشريط الأعلى، وتكاليفهم أعلى بوضوح من غيرهم في نفس العمر.

**الشريط الأسفل:** أغلبه غير مدخنين (النقاط الزرقاء)، وتكاليفهم هي الأقل وتزيد بالتدريج مع العمر.

**الشريط الأوسط:** مزيج من النقاط الحمراء والزرقاء، يعني مدخنين تكاليفهم أقل من باقي المدخنين، وغير مدخنين تكاليفهم أعلى من أغلب غير المدخنين.

### 3. Bar: متوسط التكاليف حسب التدخين وحسب المنطقة

```python
smoker_mean = df.groupby("smoker")["charges"].mean()
region_mean = df.groupby("region")["charges"].mean()

fig, axs = plt.subplots(1, 2, figsize=(12, 5))

axs[0].bar(smoker_mean.index, smoker_mean.values, color=["blue", "red"])
axs[0].set_title("Average Charges by Smoking Status")
axs[0].set_xlabel("Smoker")
axs[0].set_ylabel("Average charges")

axs[1].bar(region_mean.index, region_mean.values, color="teal")
axs[1].set_title("Average Charges by Region")
axs[1].set_xlabel("Region")
axs[1].set_ylabel("Average charges")

plt.tight_layout()
plt.show()
```

** متوسط تكاليف المدخنين أعلى بكثير من غير المدخنين (أكثر من ضعف)، بينما المناطق متقاربة نسبيًا في متوسط التكاليف.

### 4. رسمة مضلّلة عن قصد

```python
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# رسمة مضللة: المحور يبدأ من 12000
axs[0].bar(region_mean.index, region_mean.values, color="orange")
axs[0].set_ylim(12000, region_mean.max() + 200)
axs[0].set_title("Misleading: axis starts at 12,000")
axs[0].set_xlabel("Region")
axs[0].set_ylabel("Average charges")

# رسمة صادقة: المحور يبدأ من الصفر
axs[1].bar(region_mean.index, region_mean.values, color="orange")
axs[1].set_ylim(0, region_mean.max() * 1.1)
axs[1].set_title("Honest: axis starts at 0")
axs[1].set_xlabel("Region")
axs[1].set_ylabel("Average charges")

plt.tight_layout()
plt.show()
```

## Task 4: محطة أرصاد المريخ

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(2026)
sol = np.arange(1, 61)
temp = np.round(-60 + 15 * np.sin(sol / 9) + rng.normal(0, 4, 60), 1)
storm = np.zeros(60, dtype=int)
storm[[11, 12, 13, 32, 33, 46, 47, 48, 49]] = 1     # عاصفة ترابية = 1
temp[storm == 1] -= 12
wind = np.round(20 + storm * 35 + rng.normal(0, 5, 60), 1)
```

### 1. الحرارة لكل Sol (Line) وفوقها أيام العواصف بنقاط حمراء (Scatter)

```python
plt.figure(figsize=(10, 4))
plt.plot(sol, temp, label="Temperature")
plt.scatter(sol[storm == 1], temp[storm == 1], color="red", label="Storm day", zorder=3)
plt.title("Mars Temperature per Sol")
plt.xlabel("Sol (Martian day)")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```

** الحرارة تتذبذب من يوم ليوم، وفي أيام العواص (النقاط الحمراء) تنخفض الحرارة بشكل واضح عن الأيام اللي حولها.

### 2. خط المتوسط (axhline) وسهم (annotate) لأبرد يوم

```python
coldest = temp.argmin()

plt.figure(figsize=(10, 4))
plt.plot(sol, temp, label="Temperature")
plt.scatter(sol[storm == 1], temp[storm == 1], color="red", label="Storm day", zorder=3)
plt.axhline(temp.mean(), color="gray", linestyle="--", label=f"Mean = {temp.mean():.1f}")
plt.annotate(f"Coldest: Sol {sol[coldest]}, {temp[coldest]}°C",
             xy=(sol[coldest], temp[coldest]),
             xytext=(sol[coldest] - 25, temp[coldest] + 8),
             arrowprops=dict(arrowstyle="->"))
plt.title("Mars Temperature per Sol with Mean and Coldest Day")
plt.xlabel("Sol (Martian day)")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```

**ما أبرد يوم هو Sol 48 (حوالي -87.5°C) وهو يوم عاصفة، ومتوسط الحرارة حوالي -60.7°C، وأيام العواصف تنزل أكثر من غيرها عن خط المتوسط.

### 3. الحرارة والرياح في رسمتين بنفس المحور الأفقي (sharex)

```python
fig, axs = plt.subplots(2, 1, sharex=True, figsize=(10, 7))

axs[0].plot(sol, temp, color="blue")
axs[0].set_title("Temperature per Sol")
axs[0].set_ylabel("Temperature (°C)")

axs[1].plot(sol, wind, color="gray")
axs[1].scatter(sol[storm == 1], wind[storm == 1], color="red", label="Storm day")
axs[1].set_title("Wind Speed per Sol")
axs[1].set_xlabel("Sol (Martian day)")
axs[1].set_ylabel("Wind speed (km/h)")
axs[1].legend()

plt.tight_layout()
plt.show()
```

** الرياح ترتفع بشكل حاد في أيام العواصف (النقاط الحمراء)، وفي نفس الأيام تنخفض الحرارة، يعني بينهم علاقة واضحة.

### 4. أفضل نافذة إطلاق: 5 أيام متتالية بلا عواصف وأدفأ متوسط

```python
best_mean = None
best_start = None

for i in range(len(sol) - 4):
    window_storm = storm[i:i + 5]      # slicing لخمسة أيام
    window_temp = temp[i:i + 5]
    if window_storm.sum() == 0:        # بلا عواصف
        m = window_temp.mean()
        if best_mean is None or m > best_mean:
            best_mean = m
            best_start = i

print("أفضل نافذة: من Sol", sol[best_start], "إلى Sol", sol[best_start + 4])
print("متوسط الحرارة:", round(best_mean, 1), "°C")
```

**الناتج:** أفضل نافذة من **Sol 15 إلى Sol 19** بمتوسط حرارة حوالي **-46.3°C**.

```python
plt.figure(figsize=(10, 4))
plt.plot(sol, temp, label="Temperature")
plt.axvspan(sol[best_start], sol[best_start + 4], color="green", alpha=0.3, label="Best launch window")
plt.scatter(sol[storm == 1], temp[storm == 1], color="red", label="Storm day", zorder=3)
plt.title("Best Launch Window on Mars")
plt.xlabel("Sol (Martian day)")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```

**قراري:** أفضل نافذة للإطلاق هي من **Sol 15 إلى Sol 19**، بمتوسط حرارة حوالي **-46.3°C**، وهي أدفأ نافذة ممكنة بدون أي عاصفة.

**لماذا ساعدتني الرسمة أكثر من الأرقام؟** لأنها وضّحت لي بنظرة واحدة وين العواصف (النقاط الحمراء) ووين الحرارة تنزل وتطلع، فقدرت أشوف الأيام الآمنة والدافئة بسرعة، بينما لو كنت أقرأ 60 رقم ما كان سهل ألاحظ هالنمط.
