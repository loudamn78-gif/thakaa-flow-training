# تقرير اليوم الخامس عشر (Day 15) - Streamlit

### الأسئلة النظرية:

1. **كيف يعمل Streamlit؟**
   - Streamlit يعيد تشغيل ملف بايثون بالكامل من السطر الأول في كل مرة يقوم فيها المستخدم بتغيير أي قيمة أو تفاعل في الواجهة (مثل تحريك slider أو الضغط على زر).

2. **ما فائدة `@st.cache_data`؟**
   - تُستخدم لحفظ نتائج الدوال الثقيلة أو قراءة الملفات في ذاكرة التخزين المؤقت (Cache)، بحيث لا يتم إعادة تحميل الملف أو الحسابات مع كل تفاعل للمستخدم، مما يسرع أداء التطبيق بشكل كبير.

3. **لماذا يحتاج محلل البيانات تطبيقاً وليس Notebook فقط؟**
   - لأن Jupyter Notebook مخصص للتحليل البرمجي الفردي للمطور، بينما تطبيق Streamlit يحول هذا التحليل إلى لوحة تحكم تفاعلية (Dashboard) احترافية وجاهزة للاستخدام الفوري من قِبل أصحاب القرار أو العملاء دون الحاجة لمعرفة لغة بايثون.
