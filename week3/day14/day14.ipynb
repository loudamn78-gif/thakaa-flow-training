## Task 1

**1. الفرق بين Matplotlib و Seaborn:**
Matplotlib هي المكتبة الأساسية للرسم وتعطيني تحكم كامل في كل تفصيلة، لكن تحتاج كود أكثر. Seaborn مبنية فوقها، فهي أسهل وألوانها أحسن وتتعامل مع DataFrame مباشرة. نتعلم الاثنين لأن Seaborn ترسم بـ Matplotlib، فأستخدم Seaborn للرسمة نفسها و Matplotlib للعنوان والحجم والتحكم النهائي (plt.title و plt.show).

**2. وظيفة hue:**
تلوّن الرسمة حسب عمود فئوي، فتضيف بُعد جديد بدون رسمة ثانية. مثال من داتا tips: hue="time" يلوّن الفواتير حسب غداء أو عشاء. ستلوّن لونين وأكتشف هل الفرق بين الوجبتين حقيقي.

**3. الخط الأسود فوق عمود barplot:**
هذا شريط الخطأ (error bar) ويمثل فترة الثقة 95% للمتوسط. العمود يعرض المتوسط فقط، والخط يبيّن مقدار الشك فيه. خط قصير يعني المتوسط دقيق وثابت، وخط طويل يعني تشتت كبير. هو مهم لأنه يحمينا من القول «هذا اليوم أعلى» وهو ممكن بالصدفة: إذا تداخلت الخطوط بين عمودين فالفرق ممكن ما يكون حقيقي.

**4. boxplot:**
الخط داخل الصندوق = الوسيط (median). حدّا الصندوق = Q1 و Q3، والصندوق كله هو IQR (Q3 − Q1) أي الـ 50% الوسطى من البيانات. الأطراف (whiskers) تمتد إلى 1.5 × IQR، والنقاط المنفصلة خارجها هي القيم الشاذة (outliers). هذا نفس ما تعلمته الأسبوع الماضي عن IQR وقاعدة اكتشاف الشواذ.

```python
!pip install arabic-reshaper python-bidi -q

import seaborn as sns
import matplotlib.pyplot as plt
import arabic_reshaper
from bidi.algorithm import get_display

def ar(text):
    return get_display(arabic_reshaper.reshape(text))

tips = sns.load_dataset("tips")

# الرسمة 1: barplot (الخط الأسود = فترة الثقة)
sns.barplot(data=tips, x="day", y="total_bill")
plt.title(ar("متوسط قيمة الفاتورة حسب اليوم (الخط الأسود = فترة الثقة)"))
plt.xlabel(ar("اليوم"))
plt.ylabel(ar("قيمة الفاتورة"))
plt.show()

# الرسمة 2: boxplot (الوسيط، IQR، القيم الشاذة)
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title(ar("توزيع قيمة الفاتورة حسب اليوم"))
plt.xlabel(ar("اليوم"))
plt.ylabel(ar("قيمة الفاتورة"))
plt.show()
```

# Task2 التوقعات قبل التشغيل

```python
import seaborn as sns
import matplotlib.pyplot as plt
import arabic_reshaper
from bidi.algorithm import get_display

def ar(text):
    return get_display(arabic_reshaper.reshape(text))

tips = sns.load_dataset("tips")

# 1) توقعي: فواتير العشاء أكبر من الغداء
sns.histplot(data=tips, x="total_bill", hue="time")
plt.title(ar("توزيع قيمة الفاتورة حسب الوجبة"))
plt.show()
print(tips.groupby("day", observed=False)["tip"].apply(n_out))

# 2) توقعي: الأحد أعلى وسيط، والسبت أكثر قيم شاذة
sns.boxplot(data=tips, x="day", y="tip")
plt.title(ar("البقشيش حسب اليوم"))
plt.show()
print(tips.groupby("day", observed=False)["tip"].apply(n_out))

# 3) توقعي: السبت الأكثر فواتير
sns.countplot(data=tips, x="day", hue="smoker")
plt.title(ar("عدد الفواتير في كل يوم حسب التدخين"))
plt.show()

# 4) توقعي: نعم، كلما زادت الفاتورة زاد البقشيش
sns.scatterplot(data=tips, x="total_bill", y="tip", size="size", hue="time")
plt.title(ar("الفاتورة مقابل البقشيش"))
plt.show()


# 5) توقعي: الفرق بين الأيام صغير والخطوط متداخلة
sns.barplot(data=tips, x="day", y="tip")
plt.title(ar("متوسط البقشيش حسب اليوم"))
plt.show()

# 6) توقعي: total_bill و tip أقوى ارتباط
corr = tips[["total_bill", "tip", "size"]].corr()
sns.heatmap(corr, annot=True)
plt.title(ar("مصفوفة الارتباط"))
plt.show()
print(tips.groupby("day")["total_bill"].agg(["mean", "median"]).round(2))
```

```python
def n_out(s):
    q1, q3 = s.quantile([.25, .75])
    i = q3 - q1
    return ((s < q1 - 1.5*i) | (s > q3 + 1.5*i)).sum()
print(tips.groupby("day", observed=False)["tip"].apply(n_out))
```

## Task 4: التوقعات قبل التشغيل

1. countplot: أتوقع Fire الأكثر ثم Ice ثم Storm.
2. boxplot: أتوقع تنين واحد جناحه كبير جدًا عن الباقي.
3. scatterplot: أتوقع ثلاث مجموعات منفصلة حسب النوع.
4. heatmap: أتوقع wingspan_m و weight_t أقوى ارتباط.
5. القرار: أتوقع Fire للتدفئة، وIce لا يُرسل أبدًا.

```python
import numpy as np
import pandas as pd
rng = np.random.default_rng(7)
n = 150
species = rng.choice(["Fire", "Ice", "Storm"], n, p=[0.4, 0.35, 0.25])
base_wing = {"Fire": 14, "Ice": 11, "Storm": 18}
base_heat = {"Fire": 900, "Ice": 80, "Storm": 300}
dragons = pd.DataFrame({
    "dragon_id": [f"D{i:03d}" for i in range(n)],
    "species": species,
    "wingspan_m": [round(base_wing[s] + rng.normal(0, 2), 1) for s in species],
    "fire_temp_c": [round(base_heat[s] + rng.normal(0, 60)) for s in species],
    "age_years": rng.integers(20, 400, n),
})
dragons["weight_t"] = (dragons["wingspan_m"] * 0.6 + rng.normal(0, 0.8, n)).round(1)
dragons.loc[37, "wingspan_m"] = 140.0
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import arabic_reshaper
from bidi.algorithm import get_display

def ar(text):
    return get_display(arabic_reshaper.reshape(text))

# 1) countplot
sns.countplot(data=dragons, x="species")
plt.title(ar("عدد التنانين من كل نوع"))
plt.show()
print(dragons["species"].value_counts())

# 2) boxplot قبل التصحيح: نكشف التنين الغريب
sns.boxplot(data=dragons, x="species", y="wingspan_m")
plt.title(ar("طول الجناح حسب النوع (قبل التصحيح)"))
plt.show()
odd = dragons["wingspan_m"].idxmax()
print(dragons.loc[odd])
print("الوزن ÷ 0.6 =", round(dragons.loc[odd, "weight_t"] / 0.6, 1))
```

```python
# التصحيح: الوزن يدل على أن الجناح الحقيقي قريب من 14، يعني صفر زائد
dragons.loc[odd, "wingspan_m"] = round(dragons.loc[odd, "weight_t"] / 0.6, 1)
print("بعد التصحيح:", dragons.loc[odd, "wingspan_m"])

sns.boxplot(data=dragons, x="species", y="wingspan_m")
plt.title(ar("طول الجناح حسب النوع (بعد التصحيح)"))
plt.show()

# 3) scatterplot
sns.scatterplot(data=dragons, x="wingspan_m", y="fire_temp_c", hue="species")
plt.title(ar("طول الجناح مقابل حرارة النار"))
plt.show()

# 4) heatmap
num = dragons[["wingspan_m", "fire_temp_c", "age_years", "weight_t"]]
sns.heatmap(num.corr(), annot=True, fmt=".2f")
plt.title(ar("مصفوفة الارتباط"))
plt.show()

# 5) أدلة القرار
sns.boxplot(data=dragons, x="species", y="fire_temp_c")
plt.title(ar("حرارة النار حسب النوع"))
plt.show()
print(tips.groupby("day", observed=False)["tip"].apply(n_out))
```
