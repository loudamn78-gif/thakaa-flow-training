# 1. توقّع قبل أن أشغّل

| السطر | توقّعي | الناتج الفعلي | سبب الاختلاف |
|---|---|---|---|
| `type(df["score"])` | `Series` | `Series` | لا يوجد اختلاف |
| `type(df[["score"]])` | `Series` | `DataFrame` | القوسان المزدوجان `[["score"]]` يرجّعان جدولًا مو Series |
| `df.loc["b","city"]` | `Jeddah` | `Jeddah` | لا يوجد اختلاف |
| `df.iloc[1,1]` | `Jeddah` | `Jeddah` | لا يوجد اختلاف |
| `df.loc["a":"c","score"].tolist()` | `[88, 92]` | `[88, 92, 75]` | `loc` يشمل النهاية `"c"` |
| `df.iloc[0:2,2].tolist()` | `[88, 92]` | `[88, 92]` | لا يوجد اختلاف |
| `df[df["score"]>80]["name"].tolist()` | `["Sara","Noura"]` | `['Sara', 'Noura']` | لا يوجد اختلاف |
| `df["city"].value_counts()["Riyadh"]` | `2` | `2` | لا يوجد اختلاف |

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Sara", "Noura", "Reem", "Hind"],
    "city": ["Riyadh", "Jeddah", "Riyadh", "Dammam"],
    "score": [88, 92, 75, 60],
}, index=["a", "b", "c", "d"])

print(type(df["score"]))
print(type(df[["score"]]))
print(df.loc["b", "city"])
print(df.iloc[1, 1])
print(df.loc["a":"c", "score"].tolist())
print(df.iloc[0:2, 2].tolist())
print(df[df["score"] > 80]["name"].tolist())
print(df["city"].value_counts()["Riyadh"])
```

#2. الفرق بين df["score"] و df[["score"]]
`df["score"]` يرجّع **Series** (عمود واحد، بُعد واحد). `df[["score"]]` يرجّع **DataFrame** فيه عمود واحد (جدول). أستخدم Series لما أبغى قيم العمود نفسها للحساب أو الفلترة، وأستخدم DataFrame لما أبغى أحافظ على شكل الجدول أو أختار أكثر من عمود.

# 3. الفرق بين loc["a":"c"] و iloc[0:2]
`df.loc["a":"c"]` يعتمد على **اسم الفهرس** ويشمل الطرفين، فيرجّع 3 صفوف (a, b, c). `df.iloc[0:2]` يعتمد على **رقم الموقع** ولا يشمل النهاية، فيرجّع صفّين (0 و1). السبب أن loc يتعامل مع تسميات فتكون الشريحة شاملة، وiloc يتبع قاعدة بايثون العادية.

# 4. الفرق بين NumPy array و DataFrame
NumPy array مصفوفة أرقام من نوع واحد وبدون أسماء أعمدة أو صفوف. DataFrame جدول فيه أسماء أعمدة وفهرس، ويقبل أنواعًا مختلفة (نصوص وأرقام) ويتعامل مع القيم الناقصة. نحتاج Pandas لأن البيانات الحقيقية فيها نصوص وأسماء وقيم ناقصة، وPandas مبنية فوق NumPy وتسهّل التنظيف والتجميع والفلترة.

```python
trips = {
    "traveler": ["Zara", "Omar", "Lina", "Fahad", "Zara", "Nora", "Yousef", "Salma", "Khalid", "Maya"],
    "departure_year": [2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026],
    "destination_year": ["1450", "3026", "-300", "1999", "1450", "2500", "1000000", "1917", "1450", None],
    "age": [29, 35, 41, 27, 29, 33, 38, -5, 45, 31],
    "price": [12000, 30000, 45000, 8000, 12000, 22000, 999999, 15000, 12000, 9000],
    "status": ["done", "done", "cancelled", "done", "done", "DONE", "done", "done", "Done", "done"],
}
df = pd.DataFrame(trips)

# 1. EDA
print(df.isna().sum())
print("المكرر:", df.duplicated().sum())
print(df.nunique())
print(df["status"].value_counts())
```

1. EDA (قبل التنظيف)
- **القيم الناقصة:** `destination_year` فيه قيمة ناقصة واحدة (المسافرة Maya).
- **المكرر:** صف واحد مكرر (رحلة Zara إلى 1450، نفس البيانات مرتين).
- **القيم المختلفة:** `departure_year` فيه قيمة واحدة فقط (2026)، و`status` فيه 4 قيم مختلفة مع أنه المفروض 2 فقط (done وcancelled).
- **الحالة:** `value_counts` أظهر `done` (7) و`DONE` (1) و`Done` (1)، فنفس الحالة مكتوبة بثلاث طرق ولازم توحيدها.

```python
# 2. Cleaning
df = df.drop_duplicates()
df["status"] = df["status"].str.lower()
df["destination_year"] = pd.to_numeric(df["destination_year"])
df = df.dropna(subset=["destination_year"])
df = df[df["age"] > 0]
print(df.shape)

# 3. رحلة السنة 1,000,000 بسعر 999,999
print(df[df["destination_year"] == 1000000])

# 4. هل departure_year مفيد؟
print(df["departure_year"].nunique())
```

2. Cleaning
- حذفت الصف المكرر (رحلة Zara إلى 1450).
- وحّدت الحالة بـ `str.lower()` فصارت `done` و`cancelled` فقط.
- حوّلت `destination_year` إلى رقم بـ `pd.to_numeric`.
- حذفت رحلة Maya لأن وجهتها ناقصة، فرحلة بلا وجهة ما تفيد التحليل.

# 3. رحلة السنة 1,000,000 بسعر 999,999
قراري: **أحذفها**، وأعتبرها خطأ إدخال. السبب أن السعر 999,999 يشبه قيمة وهمية يكتبها الموظف لما يستعجل، والسنة مليون أكبر بكثير من بقية الوجهات (من -300 إلى 3026)، وكونها شاذة في العمودين معًا يقوّي الشك. لو كانت رحلة حقيقية لكان الأفضل التأكد من فريق الوكالة قبل الحذف.

# 4. عمود departure_year
أحذفه بـ `drop(columns=["departure_year"])`، لأن كل الرحلات 2026 (قيمة واحدة فقط)، فما يفرّق بين رحلة ورحلة، وما يفيد في التحليل أو أي موديل.

```python
# تنفيذ القرار 3
df = df[df["destination_year"] != 1000000]

# تنفيذ القرار 4
df = df.drop(columns=["departure_year"])
print(df.shape)
print(df)
```

```python
# 5. future أو past، وعدد السنوات
df["years_traveled"] = (df["destination_year"] - 2026).abs()
df["direction"] = df["destination_year"].apply(lambda y: "future" if y > 2026 else "past")

# 6. التجميع
print(df["destination_year"].value_counts())
print(df.groupby("direction")["price"].mean())

done = df[df["status"] == "done"]
print("الرحلات المكتملة:", len(done), "الإيرادات:", done["price"].sum())

longest = df["years_traveled"].idxmax()
print(df.loc[longest, ["traveler", "destination_year", "years_traveled"]])
```

5. الأعمدة الجديدة
أنشأت `years_traveled` (الفرق المطلق بين سنة الوجهة و2026) و`direction` (`future` إذا الوجهة بعد 2026، و`past` إذا قبلها).
