## Task 1

```python
import numpy as np

salaries = np.array([6000, 6500, 7000, 7200, 7500, 8000, 95000])

# 1) المتوسط عشان أتأكد من كلام صاحب الشركة
print("mean:", salaries.mean())

# 2) المتوسط والوسيط
mean = np.mean(salaries)
median = np.median(salaries)
print("mean:", mean, "| median:", median)

q1 = np.percentile(salaries, 25)
q3 = np.percentile(salaries, 75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
print("Q1:", q1, "Q3:", q3, "IQR:", iqr)
print("lower:", lower, "upper:", upper)
print("outliers:", salaries[(salaries < lower) | (salaries > upper)])

# 3) بدون المدير (آخر رقم)
no_boss = salaries[:-1]
print("mean without boss:", no_boss.mean())
print("median without boss:", np.median(no_boss))
```

1) قبل أي حساب
صاحب الشركة يقول متوسط الرواتب 19,600 ريال. الرقم صحيح حسابيًا، لكنه مضلّل،
لأن أغلب الموظفين رواتبهم بين 6000 و8000 ولا أحد منهم قريب من 19,600.
السبب إن راتب المدير 95,000 رفع المتوسط.

2) النتائج
المتوسط = 19,600 والوسيط = 7,200. الفرق كبير، وهذا يدل على وجود قيمة شاذة.
Q1 = 6750 و Q3 = 7750 و IQR = 1000، فالحد الأدنى 5250 والحد الأعلى 9250.
القيمة 95,000 أكبر بكثير من الحد الأعلى، فهي Outlier.

3) بدون المدير
المتوسط نزل من 19,600 إلى حوالي 7,033 (تغيّر كثير)، والوسيط من 7,200 إلى 7,100 (تغيّر بسيط).
أستنتج أن المتوسط حساس جدًا للقيم الشاذة، والوسيط مقاوم لها.

4) الـ Outlier ليس دائمًا خطأ
- دخل شخص مليونير في بيانات الدخل: قيمة شاذة لكنها حقيقية، ولو حذفناها نخفي الواقع.
- عملية سحب ضخمة في حساب بنكي: ممكن تكون احتيال، وهي بالضبط اللي نبي نكتشفها فلا نحذفها.
أما لو كان عمر شخص 250 سنة، فهذا خطأ إدخال ونحذفه أو نصححه.

5) فكرة z-score
الـ z-score يقول لنا كم يبعد الرقم عن المتوسط بوحدة "الانحراف المعياري".
مثل لو معدلك أعلى من متوسط الصف بكثير، يعني أنتِ بعيدة عن الباقين.
إذا كان الرقم أكبر من 3 أو أقل من -3 نعتبره غريب جدًا.

## task 3

```python
# 1) IQR للـ charges والـ bmi
def iqr_info(x):
    q1 = np.percentile(x, 25)
    q3 = np.percentile(x, 75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    out = (x < lower) | (x > upper)
    return q1, q3, iqr, lower, upper, out

for name, col in [("charges", charges), ("bmi", bmi)]:
    q1, q3, iqr, lower, upper, out = iqr_info(col)
    print(name, "| Q1:", round(q1, 2), "Q3:", round(q3, 2), "IQR:", round(iqr, 2))
    print("lower:", round(lower, 2), "upper:", round(upper, 2))
    print("outliers:", out.sum(), "| نسبتهم:", round(out.mean() * 100, 2), "%")

# 2) أعلى 5 قيم
print("top 5 charges:", np.sort(charges)[-5:])
print("top 5 bmi:", np.sort(bmi)[-5:])

# 3) z-score
z_charges = (charges - charges.mean()) / charges.std()
z_bmi = (bmi - bmi.mean()) / bmi.std()
print("z > 3 charges:", (np.abs(z_charges) > 3).sum())
print("z > 3 bmi:", (np.abs(z_bmi) > 3).sum())

# 4) متوسط التكاليف مع الـ Outliers وبدونهم
out_charges = iqr_info(charges)[5]
print("mean with outliers:", round(charges.mean(), 2))
print("mean without outliers:", round(charges[~out_charges].mean(), 2))

# 5) أتحقق من السبب الحقيقي: عمود smoker (رقم 4) اللي ما قرأناه
smoker = np.genfromtxt("insurance.csv", delimiter=",", skip_header=1, usecols=4, dtype=str)
print("smokers mean charges:", round(charges[smoker == "yes"].mean(), 2))
print("non-smokers mean charges:", round(charges[smoker == "no"].mean(), 2))
print("high charges (outliers) اللي مدخنين:", (smoker[out_charges] == "yes").mean() * 100, "%")
```

### 1) IQR
حسبت Q1 و Q3 و IQR والحدّين لعمود charges وعمود bmi.
عدد الـ Outliers في charges = ... ونسبتهم ... %، وفي bmi = ... ونسبتهم ... %.

### 2) أعلى 5 قيم
أعلى قيم في charges كبيرة جدًا مقارنة بباقي البيانات، وأعلى قيم bmi
أيضًا بعيدة عن المتوسط.

### 3) z-score
عدد القيم اللي |z| > 3 في charges = ... وفي bmi = ...
العدد يختلف عن IQR لأن z-score يعتمد على المتوسط والانحراف المعياري،
وهذي نفسها تتأثر بالقيم الشاذة، وحدّه (3) أوسع من حدّ IQR.
أما IQR فيعتمد على الربيعات وما يتأثر بالقيم المتطرفة، فيصيد عدد أكبر.

### 4) المتوسط مع الـ Outliers وبدونهم
المتوسط مع الـ Outliers = ... وبدونهم = ...
الفرق واضح، يعني الـ Outliers ترفع المتوسط.

### 5) القرار
لا، ما يمكن تكون 10% من الداتا أخطاء. نسبة كبيرة كذا ما تكون أخطاء إدخال.
فتحت الأعمدة اللي ما قرأتها ولقيت عمود smoker، ومتوسط تكاليف المدخنين

## Task 4

```python
# 1) Min-Max Normalization للعمر والـ BMI والتكاليف
def minmax(x):
    return (x - x.min()) / (x.max() - x.min())

age_n = minmax(age)
bmi_n = minmax(bmi)
charges_n = minmax(charges)

for name, col in [("age", age_n), ("bmi", bmi_n), ("charges", charges_n)]:
    print(name, "| min:", col.min(), "| max:", col.max())

# 2) بيانات العميل الأول قبل التحويل وبعده
print("before:", age[0], bmi[0], charges[0])
print("after :", round(age_n[0], 4), round(bmi_n[0], 4), round(charges_n[0], 4))

# 3) Standardization (z-score) على التكاليف
charges_z = (charges - charges.mean()) / charges.std()
print("mean ≈", round(charges_z.mean(), 6), "| std =", round(charges_z.std(), 6))

# 4) عميل جديد: عمره 40 وتكلفته 20000
# نستخدم min و max من الداتا الأصلية، مو من القيمة الجديدة
new_age = (40 - age.min()) / (age.max() - age.min())
new_charges = (20000 - charges.min()) / (charges.max() - charges.min())
print("new age scaled:", round(new_age, 4))
print("new charges scaled:", round(new_charges, 4))

# 5) كم مرة يبلغ مدى التكاليف مدى العمر
charges_range = charges.max() - charges.min()
age_range = age.max() - age.min()
print("charges range:", charges_range, "| age range:", age_range)
print("ratio:", round(charges_range / age_range, 2))
```

### 1) Min-Max Normalization
طبّقت المعادلة (x - min) / (max - min) على age و bmi و charges.
أقل قيمة في كل عمود صارت 0 وأعلى قيمة صارت 1.

### 2) العميل الأول
قبل التحويل: العمر = ... و BMI = ... والتكلفة = ...
بعد التحويل: كلها صارت بين 0 و 1.

### 3) Standardization
بعد تطبيق z-score على التكاليف، المتوسط ≈ 0 والانحراف المعياري = 1.

### 4) العميل الجديد
حوّلت قيمه باستخدام min و max من الداتا الأصلية، مو من القيمة الجديدة.
السبب: الموديل تدرّب على مقياس معيّن (الأصلي)، فلازم أي قيمة جديدة تدخل بنفس المقياس.

```python
import numpy as np

rng = np.random.default_rng(7)
brightness = 1000 + rng.normal(0, 1, 240)
for start in [30, 102, 174]:          # hidden from the astronomers!
    brightness[start:start + 5] -= 20

# 1) المتوسط والوسيط
print("mean:", round(brightness.mean(), 2), "| median:", round(np.median(brightness), 2))

# 2) z-score و mask للقراءات اللي z < -3
z = (brightness - brightness.mean()) / brightness.std()
mask = z < -3
hours = np.arange(240)
print("عدد القراءات:", mask.sum())
print("الساعات:", hours[mask])

# 3) ساعة بداية كل عبور (الساعة المنخفضة وقبلها ساعة غير منخفضة)
starts = np.where(mask[1:] & ~mask[:-1])[0] + 1
if mask[0]:
    starts = np.insert(starts, 0, 0)
print("starts:", starts)

# 4) المدة بين كل عبورين بالساعات وبالأيام
gaps = np.diff(starts)
print("gaps (hours):", gaps)
print("gaps (days):", np.round(gaps / 24, 2))
print("orbital period ≈", round(gaps.mean(), 1), "hours =", round(gaps.mean() / 24, 2), "days")

# 5) مقدار انخفاض السطوع ونسبته المئوية
normal = brightness[~mask].mean()
dip = brightness[mask].mean()
print("drop:", round(normal - dip, 2), "| %:", round((normal - dip) / normal * 100, 2), "%")
```
