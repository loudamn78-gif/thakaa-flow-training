### 1. Garbage In, Garbage Out
لو أعطيت الموديل بيانات سيئة فمخرجاته ستكون سيئة، مهما كان الموديل قويًا.
**مثال من مجالي (هوايتي الرسم):** لو دربت موديل يصنّف لوحاتي حسب الأسلوب (واقعي، كرتوني، ألوان مائية)، وبعض صوري مصنّفة بالغلط، مثل لوحة ألوان مائية مسجّلة كرتوني.
- **الخطأ الصغير:** وسم (label) خاطئ لعدد قليل من الصور، أو صور مصوّرة بإضاءة سيئة تغيّر الألوان.
- **القرار الخاطئ:** الموديل يتعلم أن خصائص الألوان المائية تعني "كرتوني"، فيصنّف لوحاتي الجديدة غلط، فأتخذ قرارات خاطئة مثل اختيار الأسلوب الأنسب لمعرضي أو لحسابي.

### 2. اكتشاف المشاكل
| المشكلة | مكانها | نوعها | قراري وسببه |
|---|---|---|---|
| قيمة مفقودة | age في الطلب 1003 | Null | أعوّضها بالوسيط (median)، لأن حذف الصف يضيع بقية بياناته |
| صف مكرر | الطلب 1002 (مرتين) | Duplicate | أحذف النسخة الثانية، لأن التكرار يضخّم وزن الصف |
| قيمة شاذة | age = 250 في الطلب 1004 | Outlier | أتحقق منها؛ إن لم أجد المصدر أعاملها كقيمة مفقودة، لأن العمر 250 مستحيل |
| عدم اتساق | city: "riyadh" و "Riyadh" | Inconsistency | أوحّد الكتابة (Riyadh)، حتى لا يعدّها الموديل مدينتين |
| نوع خاطئ | amount = "three hundred" في الطلب 1005 | Wrong type | أحوّلها إلى 300 رقمًا، لأن العمود يجب أن يكون رقميًا |

### 3. Target و Features
**المشكلة:** توقّع زحمة السير على طرق مدينتي.
- **Target:** متوسط سرعة السيارات على الطريق (كم/س)، أو مستوى الزحمة (خفيفة، متوسطة، شديدة).
- **Features:** الساعة من اليوم، يوم الأسبوع، حالة الطقس، وجود مدارس أو دوام أو مناسبة قريبة، وجود حادث أو أعمال طرق.
- **عمود لن أدخله:** مدة الرحلة الفعلية. هي متاحة في البيانات، لكنها نتيجة الزحمة نفسها وتُعرف بعد حدوثها، فإدخالها يسرّب الجواب للموديل (Data Leakage) ويعطي دقة وهمية.
