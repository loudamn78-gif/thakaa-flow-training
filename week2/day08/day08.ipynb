### 1. Garbage In, Garbage Out
لو أعطيت الموديل بيانات سيئة فمخرجاته ستكون سيئة، مهما كان الموديل قويًا.
**مثال من مجالي (هوايتي الرسم):** لو دربت موديل يصنّف لوحاتي حسب الأسلوب (واقعي، كرتوني، ألوان مائية)، وبعض صوري مصنّفة بالغلط، مثل لوحة ألوان مائية مسجّلة كرتوني.
- **الخطأ الصغير:** وسم (label) خاطئ لعدد قليل من الصور، أو صور مصوّرة بإضاءة سيئة تغيّر الألوان.
- **القرار الخاطئ:** الموديل يتعلم أن خصائص الألوان المائية تعني "كرتوني"، فيصنّف لوحاتي الجديدة غلط، فأتخذ قرارات خاطئة مثل اختيار الأسلوب الأنسب لمعرضي أو لحسابي.

### 2. اكتشاف المشاكل
| المشكلة | مكانها | نوعها | قراري وسببه |
|---|---|---|---|
| قيمة مفقودة | age في الطلب 1003 | Null | أعوّضها بالوسيط (median)، لأن حذف الصف يضيع بقية بياناته |
| صف مكرر | الطلب 1002 (مرتين) | Duplicate | أحذف النسخة الثانية، لأن التكرار يضخّم وزن الصف |
| قيمة شاذة | age = 250 في الطلب 1004 | Outlier | أتحقق منها؛ إن لم أجد المصدر أعاملها كقيمة مفقودة، لأن العمر 250 مستحيل |
| عدم اتساق | city: "riyadh" و "Riyadh" | Inconsistency | أوحّد الكتابة (Riyadh)، حتى لا يعدّها الموديل مدينتين |
| نوع خاطئ | amount = "three hundred" في الطلب 1005 | Wrong type | أحوّلها إلى 300 رقمًا، لأن العمود يجب أن يكون رقميًا |

### 3. Target و Features
**المشكلة:** توقّع زحمة السير على طرق مدينتي.
- **Target:** متوسط سرعة السيارات على الطريق (كم/س)، أو مستوى الزحمة (خفيفة، متوسطة، شديدة).
- **Features:** الساعة من اليوم، يوم الأسبوع، حالة الطقس، وجود مدارس أو دوام أو مناسبة قريبة، وجود حادث أو أعمال طرق.
- **عمود لن أدخله:** مدة الرحلة الفعلية. هي متاحة في البيانات، لكنها نتيجة الزحمة نفسها وتُعرف بعد حدوثها، فإدخالها يسرّب الجواب للموديل (Data Leakage) ويعطي دقة وهمية.

## Task 3


```python
days = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
sales = np.array([1240, 980, 1105, 1320, 1875, 2410, 2150])

# 1. المجموع، المتوسط، الوسيط، وأعلى وأقل قيمة والانحراف المعياري
print("Total:", sales.sum())
print("Mean:", round(sales.mean(), 2))
print("Median:", np.median(sales))
print("Max:", sales.max(), "| Min:", sales.min())
print("Std:", round(sales.std(), 1))

# 2. أفضل يوم وأسوأ يوم من حيث المبيعات
print("Best day:", days[sales.argmax()], sales.max())
print("Worst day:", days[sales.argmin()], sales.min())

# 3. استخدام Boolean mask لمعرفة الأيام التي مبيعاتها أعلى من المتوسط
mask = sales > sales.mean()
print("Above average:", mask)
print("Days above average:", mask.sum())
print("Their sales:", sales[mask])

# 4. مبيعات الجمعة والسبت ونسبتها من مجموع الأسبوع
weekend = sales[5:7]
weekend_total = weekend.sum()
weekend_share = (weekend_total / sales.sum()) * 100
print("Weekend total:", weekend_total, "| share:", round(weekend_share, 1), "%")

# 5. المبيعات بعد ضريبة 15%، وهدف الأسبوع القادم (زيادة 10%)
after_tax = sales * 0.85
print("After tax:", after_tax)
target = sales * 1.10
print("Next week target:", target)

# 6. استخدام np.diff لمعرفة التغير من يوم لليوم الذي يليه وفي أي يوم كانت أكبر قفزة
change = np.diff(sales)
print("Daily change:", change)
print("Biggest jump after:", days[change.argmax() + 1])
```

## Task 4


```python
branches = ["Riyadh", "Jeddah", "Dammam"]
days = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
sales = np.array([
    [1240,  980, 1105, 1320, 1875, 2410, 2150],
    [ 910,  870,  990, 1010, 1430, 1980, 2240],
    [ 640,  720,  680,  750,  990, 1210, 1080],
])

# 1. طباعة shape وشرح معناها
print("1. Shape:", sales.shape)
# المعنى: مصفوفة ثنائية الأبعاد مكونة من 3 صفوف (الفروع) و 7 أعمدة (أيام الأسبوع).

# 2. مبيعات الخميس للفرع الخميس؟ (مبيعات يوم الخميس لجميع الفروع أو فرع معين، وتحديد مبيعات الخميس ليوم الخميس لكل فرع)
print("2. Thursday all branches:", sales[:, 4])

# 3. مجموع كل فرع (axis=1) ومجموع كل يوم (axis=0) مع التعليق على الفرق
print("3. Per branch (axis=1 sum):", sales.sum(axis=1))
print("   Per day (axis=0 sum):", sales.sum(axis=0))
# الفرق: axis=1 يجمع صفوف كل فرع على حدة (إجمالي مبيعات الفرع أسبوعياً)، بينما axis=0 يجمع أعمدة كل يوم (إجمالي مبيعات كل الأيام لكل الفروع مجتمعة).

# 4. اسم أفضل فرع، واسم أفضل يوم على مستوى الشركة، ونسبة كل فرع من المبيعات
branch_totals = sales.sum(axis=1)
print("4. Best branch:", branches[branch_totals.argmax()])

day_totals = sales.sum(axis=0)
print("   Best day across company:", days[day_totals.argmax()])
print("   Share of each branch %:", np.round(branch_totals / branch_totals.sum() * 100, 1))

# 5. أعلى مبيعات في خلية واحدة: في أي فرع وأي يوم؟ باستخدام argmax و // و % من الأسبوع الأول تعيدها لصف وعمود
idx = sales.argmax()
row, col = idx // sales.shape[1], idx % sales.shape[1]
print("5. Highest single sale:", sales[row, col], "at", branches[row], days[col])
```

## Task 5


```python
import numpy as np

Depth = np.array([120, 340, 610, -1, 890, 1150, 1420, 99999, 1730, 2010,
                  2280, -1, 2560, 2840, 99999, 3120, 3390, 3650, -1, 3900])

# 1. حساب القيم التالفة (-1 و 99999) ونسبتها
bad = (Depth == -1) | (Depth == 99999)
print("Bad count:", bad.sum(), "| share:", round(bad.sum() / Depth.size * 100, 1), "%")

# 2. مقارنة المتوسط بوجود وبدون القيم التالفة
print("Mean with bad:", Depth.mean())
good = Depth[~bad]
print("Mean without bad:", good.mean())
print("Times bigger:", round(Depth.mean() / good.mean(), 1))

# 3. حساب الوسيط وأعلى قيمة بالأمتار وتحويلها إلى أقدام (Feet)
print("Median:", np.median(good))
max_m = good.max()
print("Max:", max_m, "m =", round(max_m * 3.28084, 1), "ft")

# 4. حساب الضغط ومعرفة عدد القراءات التي تتجاوز 300
pressure = good / 10 + 1
print("Pressure:", pressure)
print("Readings over 300:", (pressure > 300).sum())

# 5. حساب معدل التغير أو سرعة الهبوط باستخدام np.diff
speed = np.diff(good)
print("Descent speed:", speed)
```
